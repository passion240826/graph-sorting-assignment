# 16번

STACK을 사용한 위상 정렬 수도 코드는 다음과 같다.

1. 모든 정점의 진입 차수를 계산한다.
2. 진입 차수가 0인 정점을 STACK에 push한다.
3. STACK이 빌 때까지 4번부터 7번까지 반복한다.
4. STACK의 top에 있는 정점을 확인한다.
5. STACK에서 해당 정점을 pop하여 결과에 추가한다.
6. 해당 정점에서 나가는 모든 간선에 대해 도착 정점의 진입 차수를 1 감소시킨다.
7. 도착 정점의 진입 차수가 0이 되면 그 정점을 STACK에 push한다.
8. 결과에 모든 정점이 들어 있으면 그 결과가 위상 정렬 순서이다.

가능한 위상 정렬 순서는 모두 42가지이다.

1. 처음에는 진입 차수가 0인 정점이 a뿐이므로 a가 항상 첫 번째임
2. a 이후에는 b-c 순서와 d-e-f 순서를 지키면서 섞을 수 있음
3. b와 c가 들어갈 위치를 7자리 중 2자리 고르는 경우가 21가지이고, g와 h의 순서가 2가지이므로 21 * 2 = 42가지.

# 5번

초기 배열은 12 70 30 20 55 25이다.

## 1

선택 정렬 과정은 다음과 같다.

1. 12 70 30 20 55 25
    - 초기 배열
    - 첫 번째 위치의 12와 나머지 원소들을 비교했을 때 12가 가장 작으므로 교환하지 않음
2. 12 20 30 70 55 25
    - 두 번째 위치의 70과 30, 20, 55, 25를 비교함
    - 정렬되지 않은 영역에서 20이 가장 작으므로 70과 20을 교환함
3. 12 20 25 70 55 30
    - 세 번째 위치의 30과 70, 55, 25를 비교함
    - 정렬되지 않은 영역에서 25가 가장 작으므로 30과 25를 교환함
4. 12 20 25 30 55 70
    - 네 번째 위치의 70과 55, 30을 비교함
    - 정렬되지 않은 영역에서 30이 가장 작으므로 70과 30을 교환함
5. 12 20 25 30 55 70
    - 다섯 번째 위치의 55와 70을 비교함
    - 55가 70보다 작으므로 교환하지 않음

## 2

버블 정렬 과정은 다음과 같다.

1. 12 70 30 20 55 25
    - 초기 배열
2. 12 30 20 55 25 70
    - 인접한 두 원소를 왼쪽부터 비교함
    - 70이 30, 20, 55, 25보다 크므로 계속 오른쪽으로 교환되어 맨 끝으로 이동함
3. 12 20 30 25 55 70
    - 이미 정렬된 70을 제외하고 인접한 두 원소를 비교함
    - 55가 정렬되지 않은 영역에서 가장 크므로 다섯 번째 위치로 이동함
4. 12 20 25 30 55 70
    - 이미 정렬된 55, 70을 제외하고 인접한 두 원소를 비교함
    - 30이 25보다 크므로 교환되어 네 번째 위치로 이동함

## 3

삽입 정렬 과정은 다음과 같다.

1. 12 70 30 20 55 25
    - 초기 배열
    - 70은 12보다 크므로 그대로 둠
2. 12 30 70 20 55 25
    - 30을 앞의 정렬된 영역 12, 70과 비교함
    - 30은 70보다 작고 12보다 크므로 12와 70 사이에 삽입함
3. 12 20 30 70 55 25
    - 20을 앞의 정렬된 영역 12, 30, 70과 비교함
    - 20은 70과 30보다 작고 12보다 크므로 12와 30 사이에 삽입함
4. 12 20 30 55 70 25
    - 55를 앞의 정렬된 영역 12, 20, 30, 70과 비교함
    - 55는 70보다 작고 30보다 크므로 30과 70 사이에 삽입함
5. 12 20 25 30 55 70
    - 25를 앞의 정렬된 영역 12, 20, 30, 55, 70과 비교함
    - 25는 70, 55, 30보다 작고 20보다 크므로 20과 30 사이에 삽입함

# 6번

초기 배열은 12 70 30 20 55 25 40 50이다.

## 1

병합 정렬 과정은 다음과 같다.

1. 12 70 30 20 55 25 40 50
    - 초기 배열
    - 배열을 반으로 나누고, 나눈 배열을 정렬한 뒤 다시 병합함
2. 12 70 30 20 / 55 25 40 50
    - 왼쪽 절반과 오른쪽 절반으로 나눔
    - 왼쪽은 12 70 30 20, 오른쪽은 55 25 40 50이 됨
3. 12 70 / 30 20 / 55 25 / 40 50
    - 각 절반을 다시 두 원소씩 나눔
    - 더 작은 단위로 나누어야 병합하면서 정렬할 수 있음
4. 12 / 70 / 30 / 20 / 55 / 25 / 40 / 50
    - 원소가 하나씩 남을 때까지 나눔
    - 원소가 하나인 배열은 이미 정렬된 상태임
5. 12 70 / 20 30 / 25 55 / 40 50
    - 12와 70을 비교하여 12 70으로 병합함
    - 30과 20을 비교하여 20 30으로 병합함
    - 55와 25를 비교하여 25 55로 병합함
    - 40과 50을 비교하여 40 50으로 병합함
6. 12 20 30 70 / 25 40 50 55
    - 12 70과 20 30을 비교하며 병합함
    - 12와 20 중 12, 70과 20 중 20, 70과 30 중 30을 먼저 넣고 마지막에 70을 넣음
    - 25 55와 40 50을 비교하며 병합함
    - 25와 40 중 25, 55와 40 중 40, 55와 50 중 50을 먼저 넣고 마지막에 55를 넣음
7. 12 20 25 30 40 50 55 70
    - 12 20 30 70과 25 40 50 55를 비교하며 병합함
    - 12, 20, 25, 30, 40, 50, 55, 70 순서로 들어가 최종 정렬됨

## 2

힙 정렬 과정은 다음과 같다.

1. 12 70 30 20 55 25 40 50
    - 초기 배열
    - 오름차순 정렬을 위해 최대 힙을 만든 뒤, 가장 큰 값을 뒤로 보내는 방식을 사용함
2. 70 55 40 50 12 25 30 20
    - 부모 노드가 자식 노드보다 크도록 배열을 최대 힙으로 만듦
    - 가장 큰 값 70이 루트에 오도록 비교와 교환을 수행함
3. 55 50 40 20 12 25 30 70
    - 루트의 70과 마지막 원소 20을 교환하여 70을 정렬된 영역으로 보냄
    - 남은 원소를 다시 최대 힙으로 만들어 55가 루트가 됨
4. 50 30 40 20 12 25 55 70
    - 루트의 55와 정렬되지 않은 영역의 마지막 원소 30을 교환함
    - 남은 원소를 다시 최대 힙으로 만들어 50이 루트가 됨
5. 40 30 25 20 12 50 55 70
    - 루트의 50과 정렬되지 않은 영역의 마지막 원소 25를 교환함
    - 남은 원소를 다시 최대 힙으로 만들어 40이 루트가 됨
6. 30 20 25 12 40 50 55 70
    - 루트의 40과 정렬되지 않은 영역의 마지막 원소 12를 교환함
    - 남은 원소를 다시 최대 힙으로 만들어 30이 루트가 됨
7. 25 20 12 30 40 50 55 70
    - 루트의 30과 정렬되지 않은 영역의 마지막 원소 12를 교환함
    - 남은 원소를 다시 최대 힙으로 만들어 25가 루트가 됨
8. 20 12 25 30 40 50 55 70
    - 루트의 25와 정렬되지 않은 영역의 마지막 원소 12를 교환함
    - 남은 원소 20과 12를 비교하여 20이 루트가 됨
9. 12 20 25 30 40 50 55 70
    - 루트의 20과 마지막 원소 12를 교환함
    - 모든 원소가 오름차순으로 정렬됨

## 3

퀵 정렬 과정은 다음과 같다.

1. 12 70 30 20 55 25 40 50
    - 초기 배열
    - 각 부분 배열의 첫 번째 원소를 피벗으로 사용함
2. 12 / 70 30 20 55 25 40 50
    - 피벗 12와 나머지 원소를 비교함
    - 12보다 작은 원소가 없으므로 12는 가장 앞에 확정됨
3. 30 20 55 25 40 50 / 70
    - 남은 배열에서 피벗 70과 나머지 원소를 비교함
    - 모든 원소가 70보다 작으므로 70은 가장 뒤에 확정됨
4. 20 25 / 30 / 55 40 50
    - 30을 피벗으로 두고 20, 55, 25, 40, 50과 비교함
    - 20과 25는 30보다 작으므로 왼쪽, 55와 40과 50은 30보다 크므로 오른쪽에 둠
5. 20 25 / 30 / 40 50 / 55
    - 왼쪽 부분에서 20을 피벗으로 두면 25가 20보다 크므로 20 25가 됨
    - 오른쪽 부분에서 55를 피벗으로 두면 40과 50이 55보다 작으므로 55는 뒤에 위치함
6. 20 25 / 30 / 40 50 55
    - 40을 피벗으로 두고 50과 비교함
    - 50이 40보다 크므로 40 50 순서가 유지되고, 뒤에 55가 붙음
7. 12 20 25 30 40 50 55 70
    - 왼쪽 부분 배열, 피벗, 오른쪽 부분 배열을 합침
    - 모든 부분 배열이 정렬되어 최종 배열이 완성됨

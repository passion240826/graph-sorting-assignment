# 5번

인접 행렬 표현을 사용한다.

1. 모든 정점 사이에 간선이 있어 인접 행렬의 공간 낭비가 크지 않기 때문
2. 두 정점 사이의 간선 존재 여부를 O(1) 시간에 확인할 수 있기 때문

# 6번

인접 리스트 표현을 사용한다.

1. 실제로 존재하는 간선만 저장하므로 메모리를 효율적으로 사용할 수 있기 때문
2. 희소 그래프에서는 간선의 수가 적어 인접 리스트의 공간 복잡도 O(V + E)가 유리하기 때문
